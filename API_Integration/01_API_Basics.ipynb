## What is an API?

API stands for **Application Programming Interface**.

It is a set of rules that allows different software applications to communicate with each other.

Example:

Python program → API → Server → Data

## Web APIs

Web APIs allow applications to communicate over the internet using HTTP.

A typical flow is:

Client → HTTP Request → Server
Client ← HTTP Response ← Server

## REST API

REST (Representational State Transfer) is a common approach for building Web APIs.

REST APIs commonly use HTTP methods such as:

- GET → Retrieve data
- POST → Create data
- PUT → Update data
- DELETE → Delete data

```python
import requests

url = "https://api.github.com"

response = requests.get(url)

print(response.status_code)
print(response.text[:200])
```

## Data Engineering Relevance

APIs are an important source of data for data pipelines.

A Data Engineer may extract data from APIs, transform it, and load it into databases or data warehouses.
