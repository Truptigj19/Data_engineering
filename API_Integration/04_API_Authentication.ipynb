```python
import requests

url = "https://httpbin.org/basic-auth/user/pass"

response = requests.get(
    url,
    auth=("username", "password")
)
```

```python
headers = {
    "Authorization": "Bearer YOUR_API_TOKEN"
}

response = requests.get(url, headers=headers)
```
